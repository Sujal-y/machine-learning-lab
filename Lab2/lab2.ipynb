Problem: A delivery drone needs to travel from point A(2,3) to point B(8,11) in a 2-D coordinate 
plane. Represent the displacement as a vector, and compute its magnitude (straight-line distance) and 
unit direction vector. 

Approach: The displacement vector is simply B - A. Its magnitude (Euclidean norm) gives the 
straight-line distance the drone must travel, and dividing the vector by its own magnitude gives a unit 
vector — a vector of length 1 that represents pure direction, independent of distance. This exact 
vector arithmetic (differences, norms, and normalization) underlies distance-based algorithms like 
KNN and K-Means used in later labs. 

```python
import numpy as np 
A = np.array([2, 3]) 
B = np.array([8, 11]) 
displacement = B - A 
distance = np.linalg.norm(displacement)  #normalization
unit_vector = displacement / distance 
print("Displacement Vector:", displacement) 
print("Distance (Magnitude):", distance) 
print("Unit Direction Vector:", unit_vector)
```

Q1. 
Scenario: An e-commerce recommendation engine represents each customer's purchase pattern as a 
vector across product categories, and needs to know how 'similar' two customers are to recommend 
products. 
Task: Given two customer vectors representing spending across 4 categories, compute the cosine 
similarity between them (using dot product and norms) to quantify how similar their purchasing 
behaviour is. 

```python
cus1 = np.array([100,50,20,60])
cus2 = np.array([70,60,25,100])

#dot product
dot_product = np.dot(cus1,cus2)
print("dot:- " ,dot_product)

#finding norms
norm1 = np.linalg.norm(cus1)
norm2 = np.linalg.norm(cus2)
print("norm 1:- " ,norm1)
print("norm 2:- " ,norm2)

#cosine similarity direct formula 
cosine_similarity = dot_product / (norm1*norm2)

print(f"Cosine Similarity: {cosine_similarity:.4f}")
```

Q2. 
Scenario: A factory wants to compress its sensor readings (originally correlated across 2 machines) into 
fewer independent directions of maximum variation. 
Task: Given a small 2-feature dataset, compute the covariance matrix and its eigenvalues/eigenvectors, 
and identify the direction of maximum variance

```python
data = np.array([
    [2.5, 2.4],
    [0.5, 0.7],
    [2.2, 2.9],
    [1.9, 2.2],
    [3.1, 3.0]
])

# 2. Compute the covariance matrix
# rowvar=False treats columns as different senors (Machine 1, Machine 2)
cov_matrix = np.cov(data, rowvar=False)

# 3. Compute eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

# 4. Identify the direction of maximum variance
# Sort in descending order to get the largest eigenvalue first
sorted_indices = np.argsort(eigenvalues)[::-1]
max_idx = sorted_indices[0]

max_eigenvalue = eigenvalues[max_idx]
max_eigenvector = eigenvectors[:, max_idx]  # Columns are the eigenvectors

# Print results
print("--- Results ---")
print(f"Covariance Matrix:\n{cov_matrix}\n")
print(f"All Eigenvalues: {eigenvalues}\n")
print(f"Direction of Maximum Variance (Largest Eigenvalue = {max_eigenvalue:.4f}):")
print(f"Eigenvector Axis: {max_eigenvector}")
```

Q3.Scenario: A startup wants to fit the cheapest possible pricing function y = mx + c to just 2 data points of 
cost vs. quantity manually, before trusting Scikit-learn's regression model used later in Lab 3. 
Task: Implement Gradient Descent from scratch to fit a line y = mx + c to a small set of (x, y) points by 
minimizing the Mean Squared Error.

```python
# x = Quantity, y = Cost
x = np.array([1.0, 2.0])
y = np.array([3.0, 5.0])

# 2. Initialize parameters (slope m, intercept c)
m = 0.0
c = 0.0

# 3.parameters
learning_rate = 0.1 #step size
epochs = 500 #number of passes
n = float(len(x))  # Number of data points

print("Starting Gradient Descent...")

for epoch in range(epochs):
    # Forward pass: Compute predictions
    y_pred = m * x + c
    
    # Optional: Compute and monitor Mean Squared Error (MSE)
    mse = (1 / n) * np.sum((y - y_pred) ** 2)
    
    # Compute partial derivatives (Gradients)
    dm = (-2 / n) * np.sum(x * (y - y_pred))
    dc = (-2 / n) * np.sum(y - y_pred)
    
    # Update parameters using gradients and learning rate
    m = m - learning_rate * dm
    c = c - learning_rate * dc
    
    # Print progress every 100 epochs
    if epoch % 100 == 0 or epoch <5:
        print(f"Epoch {epoch}: Loss (MSE) = {mse:.4f}, m = {m:.4f}, c = {c:.4f}")

print("\n--- Optimization Complete ---")
print(f"Final fitted line: y = {m:.4f}x + {c:.4f}")
print(f"Target line should be: y = 2.0x + 1.0")
```

Q4.
Scenario: A call center wants to know the probability that a randomly selected call will be answered 
within 30 seconds, given historical call-answering statistics that follow a known probability distribution. 
Task: Given that call-waiting times follow a Normal distribution with mean 40s and standard deviation 
12s, use python package to compute the probability that a call is answered within 30 seconds. 

```python
# Defined parameters
mean = 40
std_dev = 12
target_time = 30

x = np.linspace(mean - (5 * std_dev), target_time, 100000)
# Calculate the height (PDF) at each point x
pdf = (1 / (std_dev * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / std_dev) ** 2)

# Calculate slice width and sum up the rectangular areas
dx = x[1] - x[0]
probability = np.sum(pdf) * dx

print(f"Probability: {probability:.4f}")
```

Q5.Scenario: A spam-filtering startup (a preview of the Naive Bayes lab) wants to reason about the
probability that an email is spam, given that it contains the word 'free', using known statistics about spam
and non-spam emails.
Task: Given P(Spam)=0.4, P('free'|Spam)=0.7 and P('free'|Not Spam)=0.1, use Bayes' theorem to
compute P(Spam | 'free' appears in the email).

```python
# Given probabilities
p_spam = 0.4
p_not_spam = 1 - p_spam
p_free_given_spam = 0.7
p_free_given_not_spam = 0.1

# Law of Total Probability
p_free = (p_free_given_spam * p_spam) + (p_free_given_not_spam * p_not_spam)

# Bayes' Theorem
p_spam_given_free = (p_free_given_spam * p_spam) / p_free

print(f"P('free') = {p_free:.2f}")
print(f"P(Spam | 'free') = {p_spam_given_free:.4f} ({p_spam_given_free * 100:.2f}%)")
```
