4.Write a Python program to group a dataset by a categorical column and compute aggregate 
sta s cs (mean, sum, count) for each group using groupby()

```python
# 1. Create a sample dataset
data = {
    "Department": ["Sales", "Engineering", "Sales", "HR", "Engineering", "Sales", "HR"],
    "Employee": ["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace"],
    "Salary": [70000, 110000, 80000, 65000, 120000, 75000, 62000],
    "Bonus": [5000, 10000, 6000, 4000, 12000, 5500, 4500],
}

df = pd.DataFrame(data)

# 2. Group by the categorical column 'Department' and aggregate 'Salary'
salary_stats = df.groupby("Department")["Salary"].agg(["mean", "sum", "count"])

# 3. Clean up the column names and formatting
salary_stats = salary_stats.rename(
    columns={"mean": "Average_Salary", "sum": "Total_Salary", "count": "Employee_Count"}
)

print("--- Salary Statistics by Department ---")
print(salary_stats)
```

5.  Write a Python program to pivot a dataset into a cross-tabulated summary table using 
pivot_table(), with a specified aggrega on func on.

```python
import pandas as pd

# 1. Create a sample sales dataset
data = {
    "Region": ["East", "West", "East", "North", "West", "North", "East", "West"],
    "Product": ["Laptop", "Laptop", "Desktop", "Tablet", "Tablet", "Laptop", "Tablet", "Desktop"],
    "Sales": [1200, 1500, 800, 400, 450, 1300, 500, 850],
}

df = pd.DataFrame(data)

# 2. Pivot the dataset: Region as rows, Product as columns, aggregate Sales with sum
pivot_summary = df.pivot_table(
    index="Region",          # Rows
    columns="Product",       # Columns
    values="Sales",          # Values to aggregate
    aggfunc="sum",           # Aggregation function (e.g., 'sum', 'mean', 'count')
    fill_value=0,            # Replace NaN where combinations don't exist
    margins=True,            # Add row/column totals
    margins_name="Total"     # Label for total rows/columns
)

print(pivot_summary)
```

6. Write a Python program to apply a custom func on row-wise/column-wise on a DataFrame 
using apply() and lambda expressions (e.g., crea ng a new derived feature).

```python
import pandas as pd

# 1. Create a sample dataset of student test scores
data = {
    "Student": ["Alice", "Bob", "Charlie", "David"],
    "Math": [85, 45, 92, 60],
    "Science": [90, 55, 88, 70],
    "English": [78, 62, 95, 65],
}

df = pd.DataFrame(data)

# 2. Row-wise apply (axis=1): Derive total score and calculate final grade
# Axis=1 passes each row as a Pandas Series to the lambda function
df["Total_Score"] = df.apply(lambda row: row["Math"] + row["Science"] + row["English"], axis=1)

df["Status"] = df.apply(
    lambda row: "Pass" if (row["Total_Score"] / 3) >= 60 else "Fail", 
    axis=1
)

# 3. Column-wise apply (axis=0): Standardize scores across the subjects
# Axis=0 (default) passes each numerical column as a Series to the lambda function
numerical_cols = ["Math", "Science", "English"]
normalized_df = df[numerical_cols].apply(
    lambda col: (col - col.min()) / (col.max() - col.min()), 
    axis=0
)

print("--- DataFrame with Row-Wise Derived Features ---")
print(df)

print("\n--- Column-Wise Scaled Scores (Min-Max: 0 to 1) ---")
print(normalized_df)
```
